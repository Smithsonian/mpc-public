# MPC Orbit Type API Tutorial

#### This tutorial explains how to use the Minor Planet Center's Orbit Type API to check the orbit types of minor planets.

The Orbit Type API helps investigate individual orbit types and is useful for checking the classification of minor planets, comets, and natural satellites.

The Obit Type API is a REST endpoint. You can use your language of choice to send POST requests to:

    https://data.minorplanetcenter.net/api/orbit-type

In the examples below we use Python code to query the API.

Further information about the Orbit Type API can be found at:
 - https://docs.minorplanetcenter.net/mpc-ops-docs/apis/orbit-type/
 - https://docs.minorplanetcenter.net/mpc-ops-docs/orbits/orbit-types/

# Import Packages
Here we import the standard Python packages needed to call the API and inspect the returned data.

````python
import requests
import json
````

````python
# Trivial function to truncate a print statement for the purposes of display
def truncated_print(s:str,l:int=2000):
    print(s[:l])
    if len(s) > l:
        print('\n(truncated for display)')
````

# Query Payload Formats

You can query the API in two ways, both of which return a JSON object with the orbit type information:

1. **Orbital parameters ONLY**:

   ```json
   {
        "a": 2.5,
        "e": 0.1,
        "i": 10.0,
        "require_mpcID": False
   }
   ```

Where `a` is the semi-major axis in AU, `e` is the eccentricity, and `i` is the inclination in degrees. `require_mpcID == False` instructs the API **not** to consider the unpacked primary provisional designation of the object at hand. `mpcID` is technically optional, but should **absolutely** be provided for comets and natural satellites, which are not uniquely defined by their orbital parameters alone.

2. **Including the `mpcID`**:

   ```json
   {
        "a": 410.50627,
        "e": 0.99928,
        "i": 129.10481,
        "mpcID": "C/2020 F3"
   }
   ```

Note that in this case, the orbit type determination is made soely on the basis of the `mpcID`. Roughly correct orbital parameters are provided for consistency.

For more information, see the [Orbit Type API documentation](https://docs.minorplanetcenter.net/mpc-ops-docs/apis/orbit-type/).

# Basic Query (JSON Response)

Here we query using a set of orbital parameters and an `mpcID` for a comet. The returned data structure is a dictionary with the following keys:

````python
url = "https://data.minorplanetcenter.net/api/orbit-type"
payload = {
    "a": 410.50627,
    "e": 0.99928,
    "i": 129.10481,
    "mpcID": "C/2020 F3"
}

response = requests.post(url, json=payload)
response.raise_for_status()

orbit_type_dict = response.json()
truncated_print(json.dumps(orbit_type_dict, indent=2))
````

#### String formated output for input that includes malformed elements.

````python
# Drop the `mpcID`, which is required as long as `require_mpcID` is not specifically set to `False`.
payload.pop('mpcID', None)

response = requests.post(url, json=payload)
truncated_print(response.text)
````

# Summary

The Orbit Type API helps to determine an orbit type for given orbital parameters and/or designations. Key points:

- **Endpoint**: `https://data.minorplanetcenter.net/api/orbit-type`
- **Method**: `POST` with a JSON payload
